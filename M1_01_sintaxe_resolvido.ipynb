### f-string

Forma padrão de interpolar. `:.2f` formata com 2 casas decimais.

````python
nome = "Ana"
total = 1234.5678

print(f"{nome} pagou R$ {total:.2f}")
print(f"Dobro: {total:.2f}")
````

### Outras formas de imprimir

`print` aceita vários argumentos separados por vírgula: ele converte tudo para
string e junta com um espaço. Não precisa de `str()` nem de `+`.

````python
nome = "Ana"
itens = 3

print("cliente:", nome, "| itens:", itens)   # vírgula converte e junta com espaço
print("a", "b", "c", sep="-")                # troca o separador
print("sem quebra de linha...", end=" ")
print("continua na mesma linha")
````

`.format()` e `%` são os estilos antigos. Você não vai escrever assim, mas vai
encontrar em código legado.

````python
total = 1234.5678

print("{} pagou R$ {:.2f}".format(nome, total))                    # posicional
print("{cli} pagou R$ {v:.2f}".format(v=total, cli="Teste"))       # nomeado
print("%s pagou R$ %.2f" % (nome, total))                          # estilo C
````

````python
# a f-string aceita o mesmo mini-formato dos exemplos acima
x = 7

print(f"{x=}")
print(f"{x:>5}|")
print(f"{x:03d}")
print(f"{0.1234:.1%}")
print(f"{1_000_000:,}")
````

### O operador morsa `:=`

Atribui **e** devolve o valor na mesma expressão. Serve para quando
você repetiria uma chamada, ou criaria uma variável só para poder testá-la.
Quase sempre precisa de parênteses.

````python
nomes = ["Ana", "Bernardo", "Cecília", "Ed"]

# sem walrus: ou chama len() duas vezes, ou cria a variável antes do if
for nome in nomes:
    if len(nome) > 5:
        print(nome, len(nome))

print("---")

# com walrus: calcula uma vez e já guarda o resultado
for nome in nomes:
    if (n := len(nome)) > 5:
        print(nome, n)
````

````python
# uso clássico: consumir algo até uma condição, sem repetir a chamada
fila = [3, 1, 4, 1, 5]

while fila and (item := fila.pop()) != 4:
    print("processando", item)

print("sobrou:", fila)
````

````python
ponto = (3, 4, 5)

print(ponto[0])
print(ponto[-1])
print(len(ponto))
print(3 in ponto)
````

### `None`: existe, mas está vazio

`None` é o "nada" do Python, o equivalente a `null`. Cuidado para não confundir
três situações diferentes: a chave **não existe**, a chave existe com valor `None`,
e a chave existe com um valor falsy como `0` ou `""`.

````python
cliente = {"nome": "Ana", "email": None}     # sabemos que ela não tem e-mail

print(cliente["email"])           # None
print(cliente.get("email"))    # chave ausente -> também None

# para distinguir os dois casos, pergunte pela chave
print("email" in cliente, "|", "telefone" in cliente)
````

````python
# Pegadinha: o padrão do .get() só entra quando a chave está AUSENTE
print(cliente.get("email", "sem email"))         # None — a chave existe!
print(cliente.get("telefone", "sem telefone"))   # aqui sim o padrão vale

# se você quer um padrão para os dois casos:
print(cliente.get("email") or "sem email")
````

````python
# Teste com `is None` (identidade), não com `== None`
if cliente["email"] is None:
    print("e-mail não cadastrado")

# None é falsy — mas 0 e "" também são. Se precisar diferenciar, seja explícito
for v in (None, 0, "", "ana@x.com"):
    print(f"{repr(v):>12}  bool: {str(bool(v)):>5}  is None: {v is None}")
````

## 8. Sets

Um set é, na prática, um dicionário só com as chaves: coleção **sem ordem** e
**sem repetição**. Duas coisas que ele faz muito melhor que lista: eliminar
duplicatas e responder "está aqui?"

````python
s = {3, 1, 2, 3, 1}

print(s)          # duplicatas somem, ordem não é garantida
print(len(s))
print(2 in s)

s.add(10)
s.discard(1)      # remove se existir; .remove() levanta KeyError se não existir
print(s)
````

````python
# Pegadinha: {} vazio é dicionário, não set
vazio = {}
print(type(vazio))

vazio = set()
print(type(vazio))
````

````python
a = {"ana", "bruno", "carla"}
b = {"bruno", "carla", "diego"}

print(a | b)   # união
print(a & b)   # interseção
print(a - b)   # diferença: está em a e não em b
print(a ^ b)   # diferença simétrica: em um ou no outro, não nos dois
````

````python
# Elementos precisam ser imutáveis (hashable) — tupla entra, lista não
ok = {(1, 2), (3, 4)}
print(ok)
````

## 9. Convertendo entre coleções

| | ordenada | mutável | duplicatas | sintaxe |
|---|---|---|---|---|
| `list` | sim | sim | sim | `[1, 2]` |
| `tuple` | sim | não | sim | `(1, 2)` |
| `set` | não | sim | não | `{1, 2}` |
| `dict` | sim (ordem de inserção) | sim | chaves únicas | `{"a": 1}` |

`list()`, `tuple()`, `set()` e `dict()` aceitam qualquer iterável. Converter é
ferramenta de trabalho: vira set para deduplicar, volta para lista para ordenar.

````python
nums = [3, 1, 2, 3, 1]

print(tuple(nums))         # lista -> tupla
print(set(nums))           # lista -> set  (perde ordem e duplicatas)
print(list(set(nums)))     # set   -> lista, já sem repetição
print(sorted(set(nums)))   # dedup + ordena: receita que você vai usar sempre

print(len(nums), "valores,", len(set(nums)), "distintos")
````

````python
# dict -> listas
pedido = {"id": 1, "valor": 150.0, "status": "aprovado"}

print(list(pedido))
print(list(pedido.values()))
print(list(pedido.items()))    # lista de tuplas (chave, valor)
````

````python
# listas/tuplas -> dict
pares = [("id", 1), ("valor", 150.0)]
print(dict(pares))              # lista de tuplas (chave, valor) -> dict

# dict -> tupla / set
print(tuple(pedido))            # as chaves numa tupla
print(set(pedido.values()))     # os valores, sem repetição
````

````python
# string <-> lista
frase = "ana,bruno,carla"

partes = frase.split(",")
print(partes)
print("-".join(partes))
print(list("abc"))
````

### Parte 5 — Cruzando com sets

14. Existe algum produto vendido por todos os quatro vendedores?

````python
# 12.

prod_ana = set()
prod_bruno = set()

for vend in vendas:
    if vend["vendedor"] == "Ana":
        prod_ana.add(vend["produto"])
    elif vend["vendedor"] == "Bruno":
        prod_bruno.add(vend["produto"])

print("só a Ana vendeu:", prod_ana - prod_bruno)
print("os dois venderam:", prod_ana & prod_bruno)

# 13.

# o `cat` do item 3 tem o "não informado" dentro, então aqui monto só as reais
cat_informadas = set()
cat_por_vendedor = {}

for vend in vendas:
    if (nome := vend["vendedor"]) not in cat_por_vendedor:
        cat_por_vendedor[nome] = set()

    # a venda 8 não tem a chave e a 17 tem valor None: as duas ficam de fora
    if (categoria := vend.get("categoria")) is not None:
        cat_informadas.add(categoria)
        cat_por_vendedor[nome].add(categoria)

print("categorias informadas:", cat_informadas)
print()

for nome, categorias in cat_por_vendedor.items():
    print(f"{nome:<6} {len(categorias)} categorias -> vendeu em todas? {categorias == cat_informadas}")

# 14.

prod_por_vendedor = {}

for vend in vendas:
    if (nome := vend["vendedor"]) in prod_por_vendedor:
        prod_por_vendedor[nome].add(vend["produto"])
    else:
        prod_por_vendedor[nome] = {vend["produto"]}

comuns = set.intersection(*prod_por_vendedor.values())

print()
print("vendidos por todos:", comuns if comuns else "nenhum")
````
