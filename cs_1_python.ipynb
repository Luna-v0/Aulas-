## Tipos e conversão

````python
int("42")                              # texto -> inteiro: 42
float("3.14")                          # texto -> float: 3.14
str(42)                                # número -> texto: '42'
bool("")                               # string vazia é falsa: False
int(3.9)                               # TRUNCA, não arredonda: 3
round(3.5)                             # arredonda para o PAR mais próximo: 4
round(2.5)                             # e por isso este dá 2, não 3
7 / 2                                  # divisão sempre float: 3.5
7 // 2                                 # divisão inteira: 3
7 % 2                                  # resto: 1
7 ** 2                                 # potência: 49
divmod(7, 2)                           # tupla (quociente, resto): (3, 1)
isinstance(42, int)                    # testa o tipo: True
````

## Strings

````python
nome = "  Ana Souza  "

nome.strip()                           # tira espaço das duas pontas
nome.strip().lower()                   # tudo minúsculo
nome.strip().upper()                   # tudo maiúsculo
nome.strip().title()                   # Primeira Letra De Cada Palavra
nome.strip().split()                   # quebra nos espaços: ['Ana', 'Souza']
"-".join(["a", "b", "c"])              # junta com separador: 'a-b-c'
nome.replace("Ana", "Bia")             # troca todas as ocorrências
"Ana" in nome                          # busca por substring: True
nome.strip().startswith("Ana")         # começa com: True
nome.strip()[::-1]                     # invertida: 'azuoS anA'
````

````python
# match/case (3.10+) compara PADRÕES: tipo, forma, conteúdo: não só igualdade
def descrever(x):
    match x:
        case 0:                        return "zero"
        case int() | float() if x < 0: return "negativo"      # tipo + condição
        case int():                    return "inteiro positivo"
        case [a, b]:                   return f"lista de 2: {a} e {b}"
        case [_, *resto]:              return f"lista maior, resto={resto}"
        case {"status": s}:            return f"dict que tem a chave status={s}"
        case str():                    return "texto"
        case _:                        return "qualquer outra coisa"

descrever(0)                           # bate no primeiro case: 'zero'
descrever(-5)                          # tipo numérico E negativo
descrever([1, 2])                      # desestrutura a lista de 2
descrever([1, 2, 3])                   # cai no padrão com *resto
descrever({"status": "ok"})            # basta a chave existir
````

````python
# and/or têm curto-circuito e devolvem um dos OPERANDOS, não True/False
None or "padrão"                       # devolve o primeiro verdadeiro: 'padrão'
"" or 0 or "achou"                     # pula os falsos: 'achou'
"a" and "b"                            # todos verdadeiros: devolve o ÚLTIMO
None and "nunca avaliado"              # o primeiro falso encerra: None
{}.get("cor") or "azul"                # padrão quando ausente OU vazio
````

## Loops

````python
nomes = ["ana", "bruno", "carla"]
notas = [4.5, 3.0, 5.0]

list(range(3))                         # [0, 1, 2]: o fim é exclusivo
list(range(2, 11, 3))                  # início, fim, passo: [2, 5, 8]
list(enumerate(nomes, start=1))        # pares (índice, valor), começando em 1
list(zip(nomes, notas))                # casa as duas listas em pares
dict(zip(nomes, notas))                # e vira dicionário direto
list(reversed(nomes))                  # de trás para frente
````

````python
saida = []
n = 0
while n < 5:
    n += 1
    if n == 2: continue                # pula o RESTO desta volta
    if n == 4: break                   # abandona o loop inteiro
    saida.append(n)

saida                                  # [1, 3]: o 2 foi pulado, o 4 parou tudo
````

````python
def tem_par(xs):
    for x in xs:
        if x % 2 == 0:
            return "achei par"
    else:                              # else do for: só roda se NÃO houve break
        return "nenhum par"

tem_par([1, 3, 5])                     # percorreu tudo sem achar: 'nenhum par'
tem_par([1, 4, 5])                     # saiu no return: 'achei par'
````

## Comprehensions

````python
nums = [1, 2, 3, 4, 5, 6]

[n * 2 for n in nums]                  # transforma cada elemento
[n for n in nums if n % 2 == 0]        # if DEPOIS do for: filtra
[n ** 2 for n in nums if n % 2]        # filtra e transforma na mesma passada
[n if n % 2 == 0 else 0 for n in nums] # if ANTES do for: substitui, não filtra
````

````python
palavras = ["sol", "python", "mar"]
matriz = [[1, 2], [3, 4], [5, 6]]

{p: len(p) for p in palavras}          # dict comprehension: chave: valor
{p[0] for p in palavras}               # set comprehension: sem repetição
[n for linha in matriz for n in linha] # dois for: ACHATA para uma lista só
[[n * 2 for n in linha] for linha in matriz]   # aninhada: mantém a forma 3x2
````

````python
sum(n ** 2 for n in nums)              # gerador: não monta a lista na memória
any(n > 5 for n in nums)               # algum atende? True
all(n > 0 for n in nums)               # todos atendem? True
max(palavras, key=len)                 # o maior segundo um critério: 'python'
sorted(palavras, key=len)              # ordena por tamanho
````

## Coleções

````python
lista = [3, 1, 2]

lista.append(4)                        # acrescenta 1 elemento no fim
lista.extend([5, 6])                   # acrescenta VÁRIOS
lista.insert(0, 0)                     # insere na posição
lista.pop()                            # remove e DEVOLVE o último
lista.count(3)                         # quantas vezes aparece
lista.index(3)                         # em que posição está a primeira ocorrência
sorted(lista, reverse=True)            # devolve nova lista ordenada, sem alterar
````

````python
lista = [0, 1, 2, 3, 4, 5]

lista[1:3]                             # do 1 ao 2: o fim é exclusivo
lista[:2]                              # do começo até o 1
lista[-2:]                             # os dois últimos
lista[::-1]                            # invertida
lista[::2]                             # de dois em dois
````

````python
d = {"a": 1, "b": 2}
d["c"] = 3                             # cria ou sobrescreve

d.get("z")                             # None em vez de KeyError
d.get("z", 0)                          # com padrão: 0
list(d.keys())                         # as chaves
list(d.values())                       # os valores
list(d.items())                        # pares (chave, valor)
{**d, "d": 4}                          # ** mescla e devolve um dict NOVO
d.setdefault("e", []).append(9)        # cria a chave com [] se não existir, e usa
````

````python
a, b = {1, 2, 3}, {3, 4}

a | b                                  # união: tudo dos dois
a & b                                  # interseção: só o que está nos dois
a - b                                  # diferença: o que só está em a
a ^ b                                  # em um ou no outro, mas não nos dois
````

````python
x, y, z = (1, 2, 3)                    # unpacking posicional
primeiro, *resto = [10, 20, 30]        # * captura o que sobrar: resto = [20, 30]
p, q = 1, 2
p, q = q, p                            # troca sem variável temporária
(p, q)                                 # (2, 1)
````

## Funções

````python
def calcular(valor: float, frete: float = 0.0, *extras, desconto: float = 0.0, **opcoes) -> float:
    """Soma valor, frete e extras, aplicando desconto."""
    return (valor + frete + sum(extras)) * (1 - desconto)

calcular(100)                          # só o obrigatório: 100.0
calcular(100, 20)                      # segundo posicional vira frete: 120.0
calcular(100, 20, 5, 5, desconto=0.1)  # *extras junta os soltos (5 e 5)
calcular(desconto=0.5, valor=100)      # nomeado dispensa a ordem: 50.0
calcular.__doc__                       # a docstring, como texto
````

````python
def estatisticas(v):
    return min(v), max(v), sum(v) / len(v)      # devolve uma TUPLA

estatisticas([10, 20, 30])             # (10, 30, 20.0)
menor, maior, media = estatisticas([10, 20, 30])    # desempacota na chamada
(lambda x: x * 2)(21)                  # lambda: função anônima de uma expressão só
````

````python
# ⚠️ o valor padrão é criado UMA vez, na definição, e sobrevive entre chamadas
def errado(item, lista=[]):
    lista.append(item); return lista

def certo(item, lista=None):
    if lista is None: lista = []       # o None força uma lista nova a cada chamada
    lista.append(item); return lista

errado("a")                            # ['a']
errado("b")                            # ['a', 'b']: vazou da chamada anterior
certo("a")                             # ['a']
certo("b")                             # ['b'], como deveria ser
````

## Erros

````python
def para_float(v):
    """Converte aceitando vírgula decimal; devolve None se não der."""
    try:
        return float(str(v).replace(",", "."))
    except (ValueError, TypeError):    # capture o específico, nunca Exception puro
        return None

para_float("45,50")                    # 45.5
para_float("n/a")                      # None, sem quebrar o programa
para_float(None)                       # None
````

````python
def dividir(a, b):
    try:
        r = a / b
    except ZeroDivisionError:
        return None
    else:                              # else: roda só se NÃO houve exceção
        return r
    finally:                           # finally: roda SEMPRE, até com return no meio
        pass

dividir(10, 2)                         # 5.0
dividir(10, 0)                         # None, tratado
````

````python
def validar(p):
    if not 0 <= p <= 1:
        raise ValueError(f"fora do intervalo: {p}")   # falhe alto e cedo
    return p

validar(0.5)                           # 0.5
# validar(1.5)                         # ValueError: fora do intervalo: 1.5
````

## Arquivos

````python
from pathlib import Path

p = Path("exemplo.txt")
p.write_text("linha 1\nlinha 2\n", encoding="utf-8")    # escreve e fecha sozinho

p.read_text(encoding="utf-8").splitlines()   # lê tudo e quebra em linhas
p.exists()                                   # o arquivo está lá? True
p.suffix                                     # a extensão: '.txt'
p.stem                                       # o nome sem extensão: 'exemplo'
p.name                                       # nome com extensão: 'exemplo.txt'
p.unlink()                                   # apaga
````

````python
p = Path("exemplo.txt")

with open(p, "w", encoding="utf-8") as f:    # with fecha mesmo se der exceção
    f.write("a,b\n1,2\n")

with open(p, encoding="utf-8") as f:
    conteudo = f.read()

conteudo                                     # 'a,b\n1,2\n'
p.unlink()
````
